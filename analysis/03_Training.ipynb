```python
from pathlib import Path
import os

project_path = Path(r"E:\AI_ML PROJECTS\Helmet Detection")

dataset_path = (
    project_path
    / "data"
    / "raw"
    / "Helmet and Number Plate Detection for Motorbike Safety.v3i.yolov8"
)

print("Dataset exists:", dataset_path.exists())
print("Dataset path:", dataset_path)
```

```python
# data.yaml file path 
yaml_path = dataset_path / "data.yaml"

print("data.yaml exists:", yaml_path.exists())
print("\n--- data.yaml content ---")

with open(yaml_path, "r") as f:
    print(f.read())
```

```python
from ultralytics import YOLO

print("Ultralytics YOLO imported successfully!")
```

```python
model = YOLO("yolov8n.pt")

print("Model loaded successfully!")
```

```python

print(model)
```

```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Training device: CPU")
```

```python
imgsz = 640
batch = 16
epochs = 50
device = "0"

print("Image size:", imgsz)
print("Batch size:", batch)
print("Epochs:", epochs)
print("Device:", device)
```

```python
results = model.train(
    data=str(dataset_path / "data.yaml"),
    imgsz=imgsz,
    batch=batch,
    epochs=epochs,
    device=device,
    workers=2,
    project=str(project_path / "results"),
    name="test_training",
    exist_ok=True
)
```
